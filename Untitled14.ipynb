Remove exact duplicate rows

```python
df = df.drop_duplicates().copy()
print("Rows after cleaning:", df.shape[0])
```

Standardize department and education names

```python
department_map = {
    "finance": "Finance",
    "finace": "Finance",
    "financ": "Finance",
    "inance": "Finance",
    "support": "Support",
    "suport": "Support",
    "suppot": "Support",
    "suppor": "Support",
    "spport": "Support",
    "upport": "Support",
    "sales": "Sales",
    "sals": "Sales",
    "saes": "Sales",
    "ales": "Sales",
    "engineering": "Engineering",
    "engineerig": "Engineering",
    "engneering": "Engineering",
    "hr": "HR",
    "h": "HR",
    "r": "HR"
}
df["department"] = (
    df["department"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map(department_map)
)
education_map = {
    "bachelors": "Bachelors",
    "bacelors": "Bachelors",
    "bachelos": "Bachelors",
    "bacheors": "Bachelors",
    "bachlors": "Bachelors",
    "bahelors": "Bachelors",
    "achelors": "Bachelors",
    "masters": "Masters",
    "masers": "Masters",
    "maters": "Masters",
    "msters": "Masters",
    "phd": "PhD",
    "ph": "PhD",
    "pd": "PhD",
    "hd": "PhD",
    "high school": "High School",
    "highschool": "High School",
    "high schol": "High School",
    "high scool": "High School",
    "high chool": "High School",
    "hig school": "High School",
    "igh school": "High School"
}
df["education_level"] = (
    df["education_level"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map(education_map)
)
print(df["department"].value_counts())
print(df["education_level"].value_counts())
```

Clean Yes/No columns and create target labels

```python

yes_no_map = {
    "yes": "Yes",
    "ye": "Yes",
    "es": "Yes",
    "no": "No",
    "n": "No",
    "o": "No"
}

for col in ["work_accident", "promotion_last_5years"]:
    df[col] = (
        df[col].astype("string")
        .str.strip()
        .str.lower()
        .map(yes_no_map)
    )
numeric_columns = [
    "left_company",
    "salary_k",
    "salary_k_dup",
    "years_at_company",
    "last_performance_score"
]
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
df["attrition"] = df["left_company"].map({
    0: "Stayed",
    1: "Left"
})

df["attrition_flag"] = df["left_company"]

print(df.head())
```

Data validation and binning

```python

df["valid_salary"] = (
    df["salary_k"].isna() |
    (df["salary_k"] > 0)
)

df["valid_tenure"] = (
    df["years_at_company"].isna() |
    df["years_at_company"].between(0, 50)
)

df["valid_performance"] = (
    df["last_performance_score"].isna() |
    df["last_performance_score"].between(1, 5)
)
df["duplicate_emp_id_flag"] = (
    df["emp_id"].notna() &
    df["emp_id"].duplicated(keep=False)
)
df["tenure_group"] = pd.cut(
    df["years_at_company"],
    bins=[-0.01, 5, 10, 15, 20, 30],
    labels=[
        "0-5 years", "6-10 years",
        "11-15 years", "16-20 years",
        "21-30 years"
    ]
)
df["salary_band"] = pd.qcut(
    df["salary_k"],
    q=4,
    labels=["Lower", "Lower-middle", "Upper-middle", "Higher"],
    duplicates="drop"
)

print(df[["tenure_group", "salary_band"]].head())
```

Outlier detection using boxplots

```python

cols = [
    "salary_k",
    "years_at_company",
    "last_performance_score"
]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, col in zip(axes, cols):
    sns.boxplot(y=df[col], ax=ax, color="lightblue")
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

Outlier detection using IQR

```python
for col in cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)

    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(col, "Possible outliers:", len(outliers))
```

Equal-width and equal-frequency binning

```python

salary = df["salary_k"].dropna()
equal_width = pd.cut(salary, bins=4)
print("Equal Width Binning:")
print(equal_width.value_counts().sort_index())
equal_frequency = pd.qcut(
    salary,
    q=4,
    duplicates="drop"
)

print("Equal Frequency Binning:")
print(equal_frequency.value_counts().sort_index())
plt.hist(
    salary,
    bins=4,
    color="orange",
    edgecolor="black"
)

plt.title("Salary Equal-Width Binning")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
edges = salary.quantile(
    [0, 0.25, 0.5, 0.75, 1]
).drop_duplicates().values

plt.hist(
    salary,
    bins=edges,
    color="skyblue",
    edgecolor="black"
)

plt.title("Salary Equal-Frequency Binning")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```
